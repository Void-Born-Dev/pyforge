```python
# Static Typing : used in other programming languages(C/C++/Java etc) that we have to tell the compiler/ interpreter about datatype we are using
#int a = 10

#Dynamic Typing : used in Python , we don't have to give data type of the variable python interpreter automatically detects the datatype of the variable 
a = 10
print(a)
```

```python
# Static Binding : used in other languages( C/C++/Java etc), a variable has the same datatype from the starting to the termination of the program

# Dynamic Binding : used in python, a  variable don't have a fixed datatype , it's datatype can change throughout the execution of the program 

b = 10 
print(type(b))
b = 6.55
print(type(b))
```

```python
# keywords(also called reserved words) : is a word that has a special, predefined meaning in a specific programming language
# 
# Identifier : In programming, an identifier is simply a unique name given by a programmer to an element in the code.
 

# Allowed Characters: Identifiers can only contain letters (a-z, A-Z), digits (0-9), and underscores (_).




# First Character Rule: An identifier cannot start with a digit. It must start with a letter or an underscore.
#1name = 'Rohan' this will give an error 
name_1  = "Rohan"
print(name_1)
_ = "Hello"
print(_)

# No Keywords: You cannot use any of Python's reserved keywords (like if, for, else, while, def) as identifiers.
# No Special Characters: You cannot use spaces or symbols like @, $, %, -, or !.

# Case-Sensitivity: Python treats lowercase and uppercase letters differently. For example, score, Score, and SCORE are three completely different identifiers.  
score = 10
Score = 30
SCORE = 60
print(id(score))
print(id(Score))
print(id(SCORE))
```

```python
# Literals : a literal is a fixed value written directly into the source code. Unlike a variable, which acts as a container whose contents can change, a literal is the actual data itself.

a = 2 # like here 2 is the literal

```

```python
#integer literals 
a = 0b1010 #binary literals
b = 200 #Decimal literals
c = 0o310 #Octal literals
d = 0x12c #Hexadecimal literals

print(a,b,c,d)#o/p : 10 200 200 300
```

```python
#float literals

float_1 = 10.5
float_2 = 1.5e2
float_3 = 1.5e-3
print(float_1, float_2, float_3)# o/p :10.5 150.0 0.0015
```
