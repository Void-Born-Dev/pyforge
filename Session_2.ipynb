```python
# In Programming, an operator is a special symbol that tells the computer to perform a specific action or calculation on data.

# In python we have following types of operator :

# Arithmetic Operator
# Relational Operator
# Logical Operator
# Bitwise Operator 
# Assignment Operator
# Membership Operator

# Arithmetic operators 
print(5+6)# Addition
print(5-6)# Subtraction
print(5*6)# Multiplication
print(5/2)# division
print(5//2)# Integer division
print(5**2)# Power of Operator

# O/P: 11
# O/P: -1
# O/P: 30
# O/P: 2.5
# O/P: 2
# O/P: 25

#Relational Operator
# return result in the boolean

print(4>5) # greater than operator
print(4<5) # greater than operator
print(4>=4) # greater or equal to operator
print(4<=4) # lesser or equal to operator
print(4==4) # Equality operator
print(4!=4) # Not equal to operator

#o/p:
# False
# True
# True
# True
# True
# False
```

```python
# Logical Operators

# and - true if all the inputs are ture or 1
# or - ture even if one input is true or 1
# not - used to switch a boolean 
# true(1) to false(0) and false(0) to true(1) 

print(1 and 0)# o/p : 0
print(1 or 0)# o/p : 1
print(not 1)# o/p : False

# Bitwise operator : 
 
print( 2 & 3)#bit wise and , o/p : 2
print( 2 | 3)#bit wise or , o/p : 3
print( ~2 ) #bit wise not , o/p : -3
print(4 >> 2)#bit wise right shift
print(5 << 2)#bit wise left shift 


# Assignment Operator

a = 2 # here '=' is the assignment operator and we are assigning value 2 to the variable a 

a+=2 
print(a)# o/p: 4

a*=2 
print(a)# o/p: 8

a/=2 
print(a)# o/p: 4

# assignment operator can be used with other operator in conjunction 

# In python increment operaotr (a++ , ++a) doesn't work

#Membership Operator 

# Membership operators in Python are used to test whether a value or a variable is found inside a sequence (like a string, list, tuple, or dictionary).

print('d' in 'delhi')#o/p : True

print('d' not in 'delhi')#o/p : False
```
