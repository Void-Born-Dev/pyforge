```python
# for loop practice questions 

# find the sum : 1/1! + 2/2! + 3/3! + ..... upto n provided by the user input

n = int(input("Enter a number:"))
fact = 1
result = 0
for i in range(1,n+1):
    fact = fact * i
    result = result + (i/fact)

print("Result:",result)
```

```python
# Nested Loop : is simply loop inside another loop

# In programming , every single time outer loop runs once, the inner loop runs completely from start to finish

# When do we use nested loop:
# Grid Systems
# Tables of data
# Combinations

# example

for i in range(1,4):
    for j in range(1,4):
        print(i,j)
```

```python
# Nested Loop problems :

# print

#   *
#   **
#   ***

n = int(input("Enter number of rows:"))#  taking number of rows as input

for i in range(1,n+1):
    for j in range(1,i+1):
        print("*",end="")
    print()
```

```python
# Nested Loop problems :

# print

#   1
#   121
#   12321


rows = int(input("Enter number of rows:"))

for i in range(1,rows+1):
   for j in range(1,i+1):
      print(j,end="")
   for k in range(i-1,0,-1):
      print(k,end="")  
   print()      
```

```python
# Loop Control Statements  : is a keyword that changes the normal flow of a loop
# 
# used to control the execution of loops
# 
# different loop control statements are :
# 
# break (The stop button)
# continue (The skip button)
# pass (The place holder)

# Break : Instantly terminates the loop completely and jumps to the next line of code outside the loop 

for i in range(1,11):
    if i == 5:
        break
    print(i)
```

```python
# application of break statement 

# WAP  to print all prime numbers present in given range 

lower = int(input("Enter a number:"))
upper = int(input("Enter a number:"))
print("Prime numbers in range ",lower,"-",upper,":")
for i in range(lower,upper+1):
    for j in range(2,i):
        if(i%j == 0):
            break
    else:
        print(i)

```

```python
# Continue : Skips the rest of the code in the current turn and jumps straight to the beginning of the next turn

for i in range(1,11):
    if i == 5:
        continue
    print(i)
```

```python
# pass : Does absolutely nothing . It is used as temporary placeholder when syntax requires a statement , but you aren't ready to write the logic yet.

for i in range(1,11):
    if i == 5:
        pass # just a place holder bcz for now we don't have logic ready 
    print(i)
```

```python
# Accessing substrings from a string :

# Extracting a character from a string :

# Indexing :

s = "hello, world"
#so , internally python assigns an index to each character in a string starting from 0

print(s[0])#o/p : h
print(s[1])#o/p : e
print(s[2])# o/p : l

# this is positive indexing 

# using index we can access characters from the string 
print(s[4])#o/p : o

#python follows concept of negative index also 

# last element (character) of string will always be accesible using -1 index irrespective of size of the string 

print(s[-1])# o/p : d
print(s[-2])# o/p : l

# indexing in negative is in following manner :

#-1: last character 
#-2: second last character 
#-3: third last character
# and so on

```

```python
# Slicing 


# here we can access more than one character from a string:

s = "hello world"

print(s[0:5])
# range : 0:5 index and last index is not included in the output
# o/p : hello

print(s[2:])# skipping second input in range means it will go upto last character of the string 
# o/p : llo world

print(s[:7])# skipping first input in range means it will start from the first character 
# o/p : hello w

print(s[:])# skipping both inputs in range means it will return whole string 
# o/p : hello world

print(s[2:9:2])# we can also pass step size here as the third input its default value is 1 
# o/p :lowr

# reverse slicing : 
# but here first input(starting value) will be greater than second (end value)

print(s[6:0:-1])# o/p : w olle

# syntax to print a string in reverse in one line 
print(s[::-1])#o/p : dlrow olleh

print(s[-5:])# o/p : world

# editing and deleting strings : 
# Python strings are immutable - so editing is not allowed 

# deletion of string
del s # del keyword is used to delete the string 
#print(s) # ---> Name error 

# we can only delete entire string not partially 
s = "hello world"

# del s[-1:-5:2] # this will give a type error
print(s)



```

```python
# functions related to the strings in Python 
# there are two type of such functions 

#1. Comman functions 
#2. String specific fucntions 


# Comman functions :
# functions which can operate on multiple data types (like list, tuple, dictionaries, etc)
# there are many such functions but we are discussing very few


# len 
# max
# min 
# sorted

# len function : counts the lenth of the strings provided as argument including spaces 
print(len("Hello World"))# o/p : 11

# max function : returns max(value) element/character in the argument
# like character with max ASCII in string
print(max('hello world'))# o/p :'w'

# min function : returns min(value) element/character in the argument
# like character with min ASCII in string
print(min('hello world'))# o/p :' '

# sorted function : returns the 'list' of sorted elements of the character of the string in case of the string 
print(sorted('hello world'))# o/p : [' ', 'd', 'e', 'h', 'l', 'l', 'l', 'o', 'o', 'r', 'w']

# the result we got was in ascending order to get the result in descending order put value of the parameter reverse = 'True'
print(sorted('hello world',reverse = True))# o/p : ['w', 'r', 'o', 'o', 'l', 'l', 'l', 'h', 'e', 'd', ' ']
```

```python
#Now functions which are only specific to the strings : 

# capitalize : using this function the first letter only of the whole string becomes capital and if any other capital letter exist in the string it will become small
s = 'hello World'
print(s.capitalize())# o/p :Hello world

# title : using this function the first letter of each word in the string becomes capital 
print(s.title())# o/p : Hello World

# upper : using this function every single letter of the string becomes capital if it's an alphabet
print(s.upper())# o/p :HELLO WORLD

# lower : using this function every single letter of the string becomes small if it's an alphabet
print(s.lower())# o/p :hello world

# swap case : using this function every letter which is small becomes capital and capital letter becomes small 
s = "Hello World"
print(s.swapcase())# O/p : hELLO wORLD
```

```python
# COUNT / FIND / INDEX 

#count funtion :  this function counts how many times a specific value(argument) appears in a string or list 
print('my name is rohan'.count('a'))# o/p : 2

#find function : this function searches a string for a specific value and return the index(postition) of it's occurance (works only on strings, not list)
print('my name is rohan'.find('n'))# o/p : 3
# if the value is not found then it returns -1

#index function : this function works same as the find function but it works on both lists and string 
# and in case of index function if the value of input is not found in the string then it gives a VALUE ERROR
print('my name is rohan'.index("n"))# o/p: 3
```

```python
# endswith/startswith

# using these function we can check if a string is starting with or ending with the argument provided or not

# endswith: 
print('my name is rohan'.endswith('an'))#o/p : True
print('my name is rohan'.endswith('e'))#o/p : False

# startswith:
print('my name is rohan'.startswith('my'))#o/p : True
print('my name is rohan'.startswith('h'))#o/p : False
```

```python
# ISALNUM/ISALPHA/ISDIGIT/ISIDENTIFIER

#isalnum : this method is used to verify if a string is alphanumeric or not 

print('yo1'.isalnum())# o/p : True
print("123".isalnum())# o/p : True
print('the'.isalnum())# o/p : True
print('hello^%'.isalnum())# o/p : False


# isalpha : this method is used to verify if all the character of a  string are alphatbets or not 

print('123'.isalpha())#o/p :False
print('hello'.isalpha())#o/p :True

# isdigit : this method is used to verify if the string contains only numbers or not 
print('123'.isdigit())#o/p :True
print('hello'.isdigit())#o/p :False

#isidentifier : gives true for valid and false for invalid identifier 
print('123'.isidentifier())#o/p :False
print('hello'.isidentifier())#o/p :True
print('first_1'.isidentifier())#o/p :True
print('f$'.isidentifier())#o/p :False
```

```python
# split/join

# split method is used to split string into words on default
# result is a list 

"is python a good lang".split()
# o/p : ['is', 'python', 'a', 'good', 'lang']
```

```python
# We can also split our string on the basis of some input 

'hi my name is nitish'.split('i')
# o/p : ['h', ' my name ', 's n', 't', 'sh']
```

```python
# Join is basically reverse of the split method 

" ".join(['hi','my','name','is','nitish'])
# o/p : 'hi my name is nitish'

# applied on a string as a method and takes a list as an argument and joins the elements of the list on the basis of the string for which it is called as a method for 
```

```python
# replace : this method is used to replace a particular substring 

'hi my name is nitish'.replace('nitish','rohan')
# here in this case 'nitish' will be replaced by the word string 'rohan'
# o/p : 'hi my name is rohan'
```

```python
# strip method : this method is used to remove the unwanted characters from the very beginning and the very end of a string .
'rohan     '.strip()
# o/p : 'rohan'
```
