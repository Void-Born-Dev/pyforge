```python
class Atm:
    #constructor (special function )-> super power 
    # to run the code inside the constructor u don't have to call the constructor explicitly becuase it operates on its own when the object of the class is created 
    # so we mostly make use of this power of the constructor by invoking such functions which are needed to be invoked at the beginning of the program like here the menu() function 
    def __init__(self):
        self.pin = ""
        self.balance = 0
        # print('Main to execute ho gya')
        self.menu()



# rules to obey while creating variables in a class for it 
# rule1 : make them inside constructor 
# rule2 : variable will start with self. then followed by its name 
# eg : self.pin

# even if you wrote 1000s of line of code in a python class it will not run if there is no object of that class using it or existing
    def menu(self):
        user_input = input("""
        Hi how can I help you ?
        1. press 1 to create pin
        2. press 2 to change pin
        3. press 3 to check balance
        4. press 4 to withdraw 
        5. Anything else to exit""")

        if user_input == '1':
            #create pin
            self.create_pin()
        elif user_input == '2':
            #change pin
            self.change_pin()
        elif user_input == '3':
            # check balance
            self.check_balance()
        elif user_input == '4':
            self.withdraw()
        else:
            pass   

    # function for creating pin 

    def create_pin(self):
        user_pin = input('Enter your pin')
        self.pin = user_pin

        user_balance = int(input('Enter balance'))
        self.balance = user_balance

        print('pin created successfully')
        self.menu()

    # function to create pin 
    def change_pin(self):
        old_pin = input('Enter old pin:')
        
        if old_pin == self.pin:
            # let him change the pin
            new_pin = input('Enter new pin')
            self.pin = new_pin
            print('pin change succcessful')
            self.menu()
        else:
            print("nai karne de sakte re baba")
            self.menu()

    # function for checking balance 
    def check_balance(self):
        user_pin = input('Enter your pin')
        if user_pin == self.pin:
            print('Your balance is :',self.balance)
        else: 
            print('Incorrect Pin')  
        self.menu() 

    # function for withdrawal
    def withdraw(self):
        user_pin = input('Enter pin: ')
        if user_pin == self.pin:
            # allow to withdraw
            amount = int(input('Enter the amount :'))
            if amount <= self.balance:
                self.balance = self.balance - amount

                print('withdrwal succesful, balance is :',self.balance)
            else:
                print('bhai tum garib ho')
        else:
            print('chori krna paap hai')
        self.menu()    
    # why self as argument passed ?
        
```
