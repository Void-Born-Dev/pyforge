Write OOP classes to handle the following scenarios:
* A user can create and view 2D coordinates
* A user can find out the distance between 2 coordinates
* A user can find find the distance of a coordinate from origin
* A user can check if a point lies on a given line
* A user can find the distance between a given 2D point and a given line

```python
class Point:
    #parameterized constructor
    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y


    # our point will look like 
    # <x,y>
    def __str__(self):
        return "<{},{}>".format(self.x_cod,self.y_cod)

    def euclidean_distance(self,other):
        return  ((self.x_cod-other.x_cod)**2 + (self.y_cod-other.y_cod)**2)**0.5


    def distance_from_origin(self):
        return ((self.x_cod)**2+(self.y_cod)**2)**0.5
        # return self.euclidean_distance(Point(0,0))
        # here in second way for performing the same operation of finding dist from (0,0) function we actually used a very imp concept ,ie we we creating class's object inside the same class 


class Line:
    def __init__(self,A,B,C):
        self.A = A
        self.B = B
        self.C = C

    def __str__(self):
        return '({})x + ({})y + ({}) = 0'.format(self.A,self.B,self.C)  

    def point_on_line(self, point):
        if self.A * point.x_cod + self.B * point.y_cod + self.C == 0:
            return "lies on the line"
        else:
            return "does not lie on the line"
    def shortest_distance(line,point):
        return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5

    def intersection(self,other):
        if self.A*other.B - self.B*other.A == 0:
            return "Lines are parallel they does not intersect"

        else:
            return "Lines intersect"        
```

```python
p1 = Point(8,0)
print(p1)
p2 = Point(1,2)
print(p2)
print(p1.euclidean_distance(p2))# dist between p1,p2
print(p2.distance_from_origin())# disctance of p1 from origin

l1 = Line(3,-4,5)
print(l1)# o/p : (3)x + (-4)y + (5) = 0
l1.point_on_line(p1)# does not lie on the line 

print(l1.shortest_distance(p1))#o/p : 5.8
print(l1.shortest_distance(p2))#o/p : 0.0

l2 = Line(1,4,-11)# o/p: (1)x + (4)y + (-11) = 0
print(l2)

l2.intersection(l1)
```

## REFERENCE VARIABLE 
* Reference variable hold the objects
* We can create objects without reference variable as well 
* An object can have multiple reference variable 
* Assigning a new reference variable to an existing object does not create a new object 

```python
# Object without a reference 

class Person:
    def __init__(self):
        self.name = 'nitish'
        self.gender = 'Male'

p = Person() # so this is the code for creating an object 
# now
# Person()# will this create an object 
# yes 
# so till now the variables which were storing object we actually storing the reference of the objects created thus these variables are called 'Reference Variable'

q = p
# so now q and p are pointing to the same object so we can say clearly now that 
#  Assigning a new reference variable to an existing object does not create a new object a single object can have multiple reference varibles

print(p.name) 
print(q.name)
q.name = 'Rohan'
print(p.name)
print(q.name) 
# since on changing value of any attribute of the variable from one reference variable lead to the change of value for every reference variable thus it is necessary for us to be very carefull while using reference variables
```

```python
# pass by reference 

class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

# outside the class hence function
# here we have provided an object ot the function as input
# also a function can returning an object 
def greet(person):
    print('Hi my name is ',person.name,'and I am a ',person.gender)
    p1 = Person('ankit','male')
    return p1

p = Person('nitish','male')
x = greet(p)  
print(x.name)# o/p : 'ankit'
print(x.gender)# o/p : 'male'

```

```python
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside the class -> function
def greet(person):
  print(id(person))
  person.name = 'ankit'
  print(person.name)

p = Person('nitish','male')
print(id(p))
greet(p)
print(p.name)

# here we are basically showing that the objext we created and the object we passed to the greet() function are same by printing their id() which comes out to be same

# So, technically when you try to pass an object to the function actually you are passing a reference of the object to that function 

```

## Object ki mutability

```python
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside the class -> function
def greet(person):
  person.name = 'ankit'
  return person

p = Person('nitish','male')
print(id(p))
p1 = greet(p)
print(id(p1))

# since the address of the p and p1 are same and objects of the class are mutable 
# because we were able to make in memory changes on the object

# we can make the objects immutable but we will talk about it later 
```

# Encapsulation

Encapsulation in object-oriented programming (OOP) is the process of bundling data (variables) and actions (methods) together into a single unit, like a class, while restricting direct access to the inner data.

```python
# what is instance variable -

# here name , country varible of the class Person are actually instance variables 
# so simply instance variable means a varible whose value is different for each instance 

class Person:

  def __init__(self,name_input,country_input):
    self.name = name_input
    self.country = country_input

p1 = Person('nitish','india')
p2 = Person('steve','australia')
```

## COLLECTION OF OBJECTS 

```python
# IF WE WANT WE CAN STORE MULTIPLE OBJECTS IN A LIST OR TUPLE OR DICTIONARY

class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender


p1 = Person('nitish','male')
p2 = Person('ankita','female')
p3 = Person('rohit','male')

l = [p1,p2,p3]
# now we can treat this list as a normal list
print(l)# this will print a list contiaining references of different object of person class 
# cause we didn't use __str__ function in the Person class 

for i in l: 
    print(i.name,i.gender)


# imp questions how can the objects be stored in the sets even if they are mutable 
    
d = {'p1':p1,'p2':p2,'p3':p3}

for k,i in d.items():
    print(k,i.name,i.gender)
```

### Static Variables (vs Instance Variables )


#### Instance Variable 
* Instance variable's values is unique for each object

**So, instance variable object ka hota ha**

* instance variables are defines inside constructor

* We use instance variable like this --> obj_name.inst_var

#### Static Variable
* Value of static variable is same for each object 

**So, static variable class ka hota ha**

* Static varibles are declared inside a class but outside any method 

* We use static variable like this --> class_name.static_var


#### Utility methods/Static Methods 
 
* These methods can be used without even creating object 
* These methods are called utility methods 
* They are accessed like this -> Class_name.Static_func()
