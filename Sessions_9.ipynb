## Aggregation(Has-A relationship)

One class owns the other class

Like there are two classes :
* Customer class 
* Address class

so, customer class has a address class
here Customer class owns the Address class

While using aggregation we cannot access private members of the owned class

We can solve this problem with the help of a getter function 

While Aggregation it is clear that we can't clearly fetch the private members of the class we own 

```python
# Aggregation Example : 

class Customer:
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address
        # here name , gender are simple attributes but address is complex to handle as it can have multiple attributs so we will make a separate class for addresss

    def print_address(self):
        # print(self.address.city,self.address.pin,self.address.state)# we cannot directly access address.city attribute beacuse it is a private member of its class 
        print(self.address.get_city(),self.address.pin,self.address.state)

    def edit_profile(self,new_name,new_city,new_pin,new_state):
        self.name = new_name
        self.address.edit_address(new_city,new_pin,new_state)    

class Address: 
    def __init__(self,city,pin,state):
        self.__city = city#  private member
        self.pin = pin
        self.state = state  

    def get_city(self):
        return self.__city 

    

    def edit_address(self,new_city,new_pin,new_state):
        self.__city = new_city
        self.pin = new_pin
        self.state = new_state



add1 = Address('gurgaon',122011,'Haryana')
#so here we are actually passing an object to a class constructor of another class and then  object is being used by the class this scenario is called Aggregation
cust = Customer('nitish','male',add1)

cust.print_address() # o/p : gurgaon 122011 Haryana  # error: Attribute Error 

# we cannot access the private members of the owned function

# Solution of this problem is a getter function 

cust.edit_profile('Rohan','Ahemdabad',111111,'Gujrat')
cust.print_address()
```

# Inheritance 

**What is Inheritance**

Inheritance in Object-Oriented Programming (OOP) is a core mechanism that allows a new class to acquire the properties and behaviors of an existing class

#### Key Terms

**Parent Class (Superclass / Base Class)**: The existing class whose features are inherited.

**Child Class (Subclass / Derived Class)**: The new class that inherits from the parent class and can add its own unique features.

**IS-A Relationship:** The logical connection created by inheritance (e.g., a Dog is an Animal). 




**What gets inherited?**

* Constructor
* Non Private Attributes
* Non Private Mehtods



**Syntax for inheriting another class** 

class Child_class(Parent_class):
  And body will be written as it is written under normal conditions 
  

**Conditions for Construtors __init__**

We know constructor is also inherited with other things from parent class in child class 

**Case 1**
If there is a constructor in child class then an object of child class will execute only constructor in child class

It will not access and use the constructor of the parent class even if it is present or not 

Thus will result in child object to be restricted from using the instance variables created by the parent constructor 

But remember even in this case the child class object will be able to access other components of the parent class which are declared outside the constructor

**Case 2**
If there is no child constructor then parent contructor will become the constructor for the child's object 


```python
class User:
    def __init__(self):
        self.name = "Nititsh"
        self.gender = "Male"
    
    def login(self):
        print('login')


# child class (Student class Inheriting form User Class)

class Student(User):

    def __init__(self):
        self.rollno = 100

    def enroll(self):
        print('enroll into the course')

u = User()
s = Student()


#print(s.name)# this will give error because name attribute is generated using parent constructor 
# and child constructor is the main constructor of child

print(s.rollno)
s.enroll()
s.login()# method of parent class accessed by child's object

```

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def __init__(self, price, brand, camera):
        print ("Inside Smartphone phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a smartphone")

s=SmartPhone(20000, "Apple", 13)

# her this example illustrates that the constructor as method is overriden so basically it is an example of method overriding 
# but, since the constructor is being overriden we will also call it 'Constructor Overriding'
```

```python
# Super keyword - super keyword use karke aap parent ki cheezo ko call kr sakte hoo
# super is way to access parent key methods 

class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        super().buy()

s=SmartPhone(20000, "Apple", 13)

s.buy()


```

```python
# super -> constuctor
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside smartphone constructor')
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
        print ("Inside smartphone constructor")

s=SmartPhone(20000, "Samsung", 12, "Android", 2)

print(s.os)
print(s.brand)
# here we triggered parent's constructor using child's constructor 

# So, using super we created a smart way to call the constructor of both child and parent 



# can super access parent ka data?
# using super outside the class  
# No, we can't use Super keyword outside the class and we use it inside the child class 
# using super keyword outside the class gives RuntimeError


```

```python
# single inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
```

```python
# multilevel
class Product:
    def review(self):
        print ("Product customer review")

class Phone(Product):
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()# here we are able to access the member of parent class
s.review()# also , we are able  to access the member of grandparent class

# the chain of classes in both directions(child side and parents side) can be more an more means we can attach classes above parent from which parend will and inherit and we can attach a class below child class which will inherit from child 

```

```python
# Hierarchical
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

class FeaturePhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
FeaturePhone(10,"Lava","1px").buy()
```

```python
# Multiple
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def review(self):
        print ("Customer review")

class SmartPhone(Phone, Product):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

```python
# the diamond problem

class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
# here in this example multiple inheritance is applied so we are able to use buy method of both the parents which is a method avialable in both class so this causes ambiguity that which buy function will be called it is decided by the which class was inherited first 
# like in this example it was SmartPhone(Phone,Product):
# so when we will call buy method using child class it will call the buy method in the Phone class 
# this behavior is specific in Python 

# This concept is called MRO - Method resolution order 
# And python uses MRO to select what to do in this scenario  
```

### Polymorphism

- Method Overriding
- Method Overloading
- Operator Overloading


Polymorphism is the concept where a single function, method, or operator can behave differently depending on the type of data or object it is interacting with. 


**Method Overriding :** Method overriding happens when a child class redefines a function that it inherited from its parent class to give it a different, more specific behavior.


**Method Overloading :** Method overloading is when a class has multiple functions with the exact same name, but different inputs (different types or number of parameters). The program automatically knows which one to use based on the data you pass into it.


**Operator Overloading :** Operator overloading allows built-in symbols (like +, -, or *) to behave differently depending on the data type they are used with.

```python
#  Method Overriding
```

```python
# Method Overloading

class Shape:
    def area(self,radius):
        return 3.14*radius*radius

    def area(self,l,b):
        return l*b

    # Method overloading makes our code look clean     

circle = Shape()
Rect = Shape()

# print(circle.area(5))# this will give error
print(Rect.area(10,20))# o/p : 100
  
# method overloading will not work in this way in python cause in the class written the python will only consider the latest version of the area() method as the code of area() method    
```

```python
# so using following way we can apply method overloading in python actually

class Shape:

  def area(self,a,b=0):
    if b == 0:
      return 3.14*a*a
    else:
      return a*b

s = Shape()

print(s.area(2))
print(s.area(3,4))
# we use default variable concept this is the smart way 
# so we can also say traditional method overloading does not work in Python.
```

```python
# Operator overlaoding - same operator work differently on the basis on the type of the opperands 

# eg : 

print('hello'+'world')# here '+' operator is performing concatination
print(5+7)# here it is performing addtion
print([1,2,3]+[4,5])# here it is merging the list 

# this is a simple example of operator overloading 
```

## ABSTRACTION 

In computer science, abstraction means hiding the complex internal details of how something works and only showing the essential features needed to use it.

Abstract Class in python is any class which contain atleast one abstract method and it inherits 'ABC'
Also there can be multiple abstract methods can also be stored in a single abstract class
We can make the object of the abstract class 

* TYPES OF METHOD IN PYTHON 
1. Concrete method 
2. Abstract method

Abstract method is a method in which nothing is implemented or written 





abc module (abstract base classes moduel )

ABC - ABSTRACT BASE CLASS
abstract method 

###

```python
from abc import ABC,abstractmethod

class BankApp(ABC):

    def database(self):
        print('connected to database')

    # this is an abstract mehtod decorator which creates a method abstract method 
    @abstractmethod
    def security():
        pass
```

```python
class MobileApp(BankApp):
    def mobile_login(self):
        print('login into mobile')

    def security(self):
        print('mobile security')     
```

```python
#bank = BankApp()
# TypeError
# this will give error cause we can make instance or object of an abstract class

mob = MobileApp()

#TypeError: Can't instantiate abstract class MobileApp without an implementation for abstract method 'security'
# this error will pop up if we try to make child class without the abstract class 'security'
# once we have create the absctract function's implementation in the child class it will not give this error 
mob.database()
mob.mobile_login()
mob.security()
# now we can access any method 

# now in real workld we may come accross some scenarios in which we have to make the child class to do some favours of tasks before accessing the members of the parent class

# like implementing a security method in bank application's different versions 

# so like samna wale ko kuch apse lene ke badle pahle appka clause agree karna padega
```
