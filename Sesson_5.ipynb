# Tuple
  A tuple in Python is  similar to a list . The difference b/w the two is that we cannot change the elements of a tuple once it is assigned where as we can change the elements of a list .

  In short, a tuple is an immutable list . A tuple cannot be change in any way once it is created .

### Charracteristics 
  Ordered
  Immutable
  Allows duplicate  
  Uses parentheses for its structure

```python
# Creating tuples : 

# creating empty tuple : 
t1 = ()
print(t1)# o/p: ()

# creating tuple with a sing element 
t2 = ('hello')#wrong way this will create a string 
print(t2)# o/p :'hello'
print(type(t2))# o/p :<class 'str'> 

t2 = ('hello',)# now this will create a tuple 
print(t2)# o/p :('hello',)
print(type(t2))# o/p :<class 'tuple'>

# homogeneous tuple 
t3 = (1,2,3,4)
print(t3)# o/p :(1,2,3,4)

# heterogeneous tuple 
t4 = (1,2.5,True,[1,2,3])
print(t4)#o/p :(1,2.5,True,[1,2,3]) 

# 2-D tuple 
t5 = (1,2,3,(4,5))

# using type conversion 
t6 = tuple('hello')
print(t6)# o/p :('h', 'e', 'l', 'l', 'o')
```

```python
# Accessing Items from a tuple : 

t1 = (1,2,3,4)
# indexing
print(t3[0])
print(t3[-3])
# similar like list 

# slicing
print(t3[0:4])# o/p :(1,2,3,4)
print(t3[0:4:2])# o/p :(1,2)
print(t3[-3:-1])# o/p :(2,3)
print(t3[::-1])# o/p :(4,3,2,1)

# accessing element in 2d list 
t2 = (1,2,3,(4,5))
print(t2[3][0])# o/p :4
# works similar like list 
```

## Operations on the tuple
   Arithmetic(+ and *) operations 
   membership operations 
   iteration

```python
# Arithmetic operations 
t1 = (1,2,3,4)
t2 = (5,6,7,8)

# + this will concatenate/merge the operand tupl3es
print(t1+t2)# o/p: (1,2,3,4,5,6,7,8)

# * 
print(t2*3)# this will same as it worked with list
# o/p : (5,6,7,8,5,6,7,8,5,6,7,8)
```

```python
# membership operations 

t = (1,2,3)

print(1 in t)# o/p :True
print(1 not in t)# o/p :False

# iteration

for i in t : 
    print(i,end = " ")# o/p :1 2 3 
```

# Sets 
 
  A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable ( cannot be changed). However, a set is itself is mutable , we can add or remove items from it. 

  Set can also be used to perform mathematical set operations like union , intersection , symmetric differences , etc.

  Characteristics :

  **Ordered**
  **Mutable**
  **No duplicates**
  **Can't contain mutable datatypes**

  

```python
# Creating sets : 

#empty set
s = {} # wrong way because this will create an empty dictionary
print(type(s))# o/p :<class 'dict'>
s = set()# this will work
print(type(s))## o/p :<class 'set'>

# 1D and 2D set
s1 = {1,2,3}
print(s1)# o/p :{1,2,3}

# s2 = {1,2,{1,23}}# this will give a TypeError 
# so you cannont have a multidimensional set cause we can't keep mutable data type items in sets 

# homo and hetero set

# homo 
s1 = {1,2,3}
# hetero
s2 = {1,'hello',4.5,True}
print(s2)# o/p : {1,4.5,'hello}
# here in o/p we can see that even if we added 4 items in the set during declaration then also it contians only 3 elements why
# cause : duplicates are not allowed in set and 'True and 1 are connsidered equal in python'

# sets are unordered and how it's data is stored is decided using that is working in backsid 

# creating set using type conversion 
s4 = set([1,2,3])
print(s4)# o/p :{1,2,3}

# duplicates are not allowed 
s5 = {1,2,2,3,3,1}
print(s5)# o/p : {1,2,3}

# set can't have mutable items 
#s6 = {1,2,[1,3]}# this will give a TypeError
```

   Since, sets are unordered, we can apply indexing and slicing on the sets thus,
   Accessing items and editing items in a set is not allowed 


```python
# Adding items in a set : 

# add  : this method us used to add an individual item in set but the position of the item in set will be not known to you 
s = {1,2,3,4}
print(s)# o/p :{1,2,3,4}
s.add(5)
print(s)# o/p :{1,2,3,4,5}
# hashing controls the item's position in a set


# update : used to add multiple items in a set by passing a list of items to be added in this method 
s.update([6,7,8])
print(s)# o/p : {1,2,3,4,5,6,7,8}
```

```python
# deleting items :

s = {1,2,3,4,5}
print(s)# o/p {1,2,3,4,5}

# del : deletes entire list 
del s 
#print(s) this will give a NameError now 

s = {1,2,3,4,5}

#discard : deletes item using its value 
s.discard(5)
print(s)# o/p : {1,2,3,4}
# if we try to delete an item from a set which doesn't exist in this set then it will not give any error also it will not delete anything from the set 

# remove : works same as the discard method only differs in behavior if we try to delete anything which is not present in the set .
# if we try to delete an item from set which is not present in int it will throw key error
s.remove(4)
print(s)# o/p : {1,2,3}
# s.remove(50) will give KeyError

# pop method will randomly delete any item form the set 
s.pop()
print(s) # now if we print the set we will see any random item is missing from it

# clear : this method when applied on set will make set empty 
s.clear()
print(s)# o/p : set() --> empty set 
```

## FUNCTIONS ON A SET 
   Union (|)
   Intersection (&)
   Difference (-)
   Symmetric Difference (^)
   Membership operator (in, not in)
   Iteration (looping)
   

```python
# Set operations
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# Union (|)
print(s1 | s2)# o/p : {1,2,3,4,5,6,7,8}

# Intersection (&)
print(s1 & s2)# o/p : {4,5}

# Difference (-)
print(s1 - s2)# o/p : {1,2,3}
print(s2 - s1)# o/p : {6,7,8}

# Symmetric Difference 
print(s1 ^ s2)# o/p : {1,2,3,6,7,8}

# Membership operators 
print(1 in s1)#o/p : True 
print(1 not in s1)#o/p : False

# Iteration 
for i in s1:
    print(i,end=" ") 
# o/p :1 2 3 4 5    
```

```python
# Set functions and methods 

# len/sum/min/max/sorted
s1 = {1,2,3,4,5}
print(len(s1))#o/p :5
print(sum(s1))#o/p :15
print(max(s1))#o/p :5
print(min(s1))#o/p :1
print(sorted(s1))#o/p :[1,2,3,4,5]
print(sorted(s1,reverse=True))#o/p :[5,4,3,2,1]
# also , here sorted returns the result in the form of a list 
```

```python
#union/update 

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
# union function works same as like the statement 's1 | s2'
print(s1.union(s2))# o/p:{1, 2, 3, 4, 5, 6, 7, 8} 
s1.update(s2)# this method will save the content of s1 union s2 in the s1 
print(s1)# o/p :{1,2,3,4,5,6,7,8}
print(s2)# o/p :{4,5,6,7,8}
```

```python
# intersection / intersecton_update

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
# intersection function works same as like the statement 's1 & s2'
print(s1.intersection(s2))# o/p:{4,5} 
s1.intersection_update(s2)# this method will save the content of s1 intersection s2 in the s1 
print(s1)# o/p :{4,5}
print(s2)# o/p :{4,5,6,7,8}
```

```python
# differnece/difference_update

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
# difference function works same as like the statement 's1 - s2'
print(s1.difference(s2))# o/p:{1,2,3} 
s1.difference_update(s2)# this method will save the content of s1 difference s2 in the s1 
print(s1)# o/p :{1,2,3}
print(s2)# o/p :{4,5,6,7,8}
```

```python
# symmetric difference / symmetric difference update

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
# symmetric function works same as like the statement 's1 ^ s2'
print(s1.symmetric_difference(s2))# o/p:{1,2,3,6,7,8} 
s1.symmetric_difference_update(s2)# this method will save the content of s1 symmetric difference s2 in the s1 
print(s1)# o/p :{1,2,3,6,7,8}
print(s2)# o/p :{4,5,6,7,8}
```

```python
# disjoint sets : sets which don't share any item comman 

s1 = {1,2,3,4,5}
s2 = {6,7,8}
s3 = {4,5}

# here s1,s2 are disjoint sets 


print(s1.isdisjoint(s2))# o/p : True
print(s1.isdisjoint(s3))# o/p : False

# and s1 and s3 are not disjoint and also s3 is a subset of the s1 and s1 is superset of s1

print(s3.issubset(s1))# o/p: True
print(s1.issuperset(s3))# o/p: True
```

```python
# copy function : creates a shallow copy of the og set 

s1 = {1,2,3}
s2 = s1.copy()

print(s1)
print(s2)
print(id(s1))
print(id(s2))

# as we can see both s1 and s2 will the have the same value but will have different addresses
```

# PYTHON DICTIONARIES 

   In python dictionary is a collection of keys and values, used to stroe date values like a map unlike other data types which store single value as an element

   In some languages it is known as map or associate arrays 

   basic example of dictionary in python is as follows 

   dict = {
        'name': 'nitish',
        'age': 33,
        'gender' : 'male'
        }


    Characteristics : 

    Mutable 
    Indexing has no meaning
    Keys can't be duplicated 
    Keys can't be mutable items 

```python
# creation a dictionary 

# empty dictionary 
d = {}

# 1-D dictionary
d1 = {
    'name': 'Nitish',
    'gender': 'male',
}

# with mixed values 
d2 = {
    (1,2,3): 1,
    'hello': 'world'
}

# 2-D dictionary 
s = {'name': 'nitish',
     'college': 'bit',
     'sem': 4,
     'subjects':{
         'dsa': 50,
         'maths': 43,
         'eng': 64
     }
     }

# using sequence dictionary function 
d4 = dict([(1,1),(2,2),(3,3)])
# argument for this function : list of tuples 
print(type(d4))# o/p :<class 'dict'>

# duplicatekeys --> not allowed 
d5 = {
    'name': 'Rahul',
    'name': 'Rocky'
}
print(d5)# o/p : {'name': 'Rocky'}

# multiple items as keys not allowed 
# d6 = {'name': 'nitish',[1,2,3]:2} this will give a TYPEERROR
# BCZ in this list was a mutable an it is not allowed for a key to be mutable entity
```

```python
# Accesing items from a dictionary 
# dictiories are ordered but in the order of insertion 

my_dict = {
    'name': 'jack',
    'age': 26
}

# using values using key 
print(my_dict['name'])# o/p : 'jack'

# using get function 
print(my_dict.get('age'))# o/p : 26
```

```python
# Adding new key value pairs

d1 = {
       'name': 'jack',
       'age': 32,
        3:3
       }


d1['new_key']= 'new_value'
# here 'new_key' is the key of the new item and 'new_value' is the new value to be added in the dictionary d1 
print(d1)# o/p :{'name': 'jack', 'age': 32, 3: 3, 'new_key': 'new_value'} 
```

```python
# Removing key value pair from dictionary
d1 = {
       'name': 'jack',
       'age': 32,
        3:3,
        'new_key': 'new_value'
       }
# pop()
# we can delete the key value pair of our choice using this function by passing key of that pair as argument in this method

d1.pop(3)# deletes the corresponding key value pair

print(d1)# o/p : {'name': 'jack', 'age': 32, 'new_key': 'new_value'}

# popitem() deletes last key value pair from the dictionary 
d1.popitem()

print(d1)# o/p :{'name': 'jack', 'age': 32}

# del using del keyword we can delete entire dictionary and also delete a specific key value pair using followinf but we have to pass the key of that pair in following 
# del d['key']#eg
del d1['age']

print(d1)# o/p :{'name': 'jack'}

# clear method can be used to make dictionary empty 
d1.clear()

print(d1)# o/p :{}

del d1 # this will delete the d1 dictionary 
# print(d1) # this will give NameError since d1 dictionary is deleted before this statement 
```

```python
# Editing key value pair 
d = {
    'name':'Nitish',
    'age':29,
    'key':'old_value'
}
d['key'] = 'new_value'# we can directly access and edit the value of the specific key value pair using this way
print(d)# o/p:{'name': 'Nitish', 'age': 29, 'key': 'new_value'} 
```

```python
# Dictionary operations 

# membership
# iterartion 

s = {'name': 'Rahul',
     'college': 'bit',
     'sem': 3,
     'subjects':{
         'dsa': 50,
         'maths': 43,
         'eng': 64
     }
     }

#pass key in here in place for now key is written you will find if this key exist or not in the list
print('key' in s )# o/p : False

# don't check using value 
# cause membership operator works on keys

print('name' in s )# o/p : True 
```

```python
# iteration

d = {
    'name':'Rahul',
    'gender':'male',
    'age':36
}

# for loops

for i in d:
    print(i,end=' ')
# o/p:name gender age    
# this will only iterate through the keys in the dictionary 

print()

for i in d:
    print(i,d[i])
# o/p :  name Rahul
#        gender male
#        age 36  

# iterating through the .item method
for k,i in d.items():
    print(k,i)
# o/p :
# name Rahul
# gender male
# age 36    
```

```python
# Dictionary functions :

# len() - give the number of key-value pairs in dictionary 
# sorted(d) sorts the keys -> returns in the list 
# sorted(d,reverse=True) sorts the keys -> returns in the list in descending order 

# max min will also work in dictionary but when all the value are numeric or similar (not clear) 

d = {
    'name':'Nitish',
    'gender':'male',
    'age':33
}

# items 
print(d.items())# o/p :dict_items([('name','nitish'),('gender','male'),('age',33)])
# returns key value pairs in tuple in a list 

print(d.keys())
# dict_keys(['name', 'gender', 'age'])

print(d.values())# o/p : dict_values(['nitish','male',33])
```
